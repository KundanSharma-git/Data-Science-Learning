```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

#Central Limit Theorem

```python
pop_data = [np.random.randint(10,100) for i in range(10000)]
pop_table = pd.DataFrame({"Pop_data": pop_data})
```

```python
pop_table
```

```python
sns.kdeplot(x="Pop_data",data=pop_table)
plt.figure(figsize=(3,2))
plt.show()
```

```python
sam_mean = []
for no_sample in range(100):
    sample_data = []
    for data in range(500):
        sample_data.append(np.random.choice(pop_data))
        sam_mean.append(np.mean(sample_data))
```

```python
sample_M = pd.DataFrame({"Sample Mean": sam_mean})
sample_M
```

```python
sns.kdeplot(x="Sample Mean",data = sample_M)
plt.show()
plt.figure(figsize = (2,1))
```
